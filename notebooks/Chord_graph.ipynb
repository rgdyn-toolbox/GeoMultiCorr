```python
import pandas as pd
import holoviews as hv
from holoviews import opts, dim
from itertools import combinations, permutations
from bokeh.models import FixedTicker

hv.extension("bokeh")

# -----------------------------
# 1. Dates
# -----------------------------
dates = [
    "2000-03-01",
    "2001-03-01",
    "2002-03-01",
    "2003-03-01",
    "2004-03-01",
    "2005-03-01",
    "2006-03-01",
    "2007-03-01",
    "2008-03-01",
    "2009-03-01",
    "2010-03-01",
]

# Convert and sort dates ascending
dates = sorted(pd.to_datetime(dates))
date_labels = [d.strftime("%Y-%m-%d") for d in dates]

# -----------------------------
# 2. Choose pair logic
# -----------------------------

# Undirected pairs: A-B only once
pairs = list(combinations(date_labels, 2))

# Directed permutations: A-B and B-A
# pairs = list(permutations(date_labels, 2))

# -----------------------------
# 3. Create node table
# -----------------------------
date_to_id = {date: i for i, date in enumerate(date_labels)}

nodes = pd.DataFrame({
    "index": list(date_to_id.values()),
    "date": date_labels,
    "date_order": list(date_to_id.values())
})

# -----------------------------
# 4. Create edge table
# -----------------------------
edges = pd.DataFrame([
    {
        "source": date_to_id[source],
        "target": date_to_id[target],
        "value": 1
    }
    for source, target in pairs
])

# -----------------------------
# 5. Optional: make colorbar labels show dates
# -----------------------------
def date_colorbar_hook(plot, element):
    colorbar = plot.handles.get("colorbar", None)

    if colorbar is not None:
        colorbar.title = "Date"
        colorbar.ticker = FixedTicker(ticks=list(date_to_id.values()))
        colorbar.major_label_overrides = {
            i: date for date, i in date_to_id.items()
        }

# -----------------------------
# 6. Build chord diagram
# -----------------------------
chord = hv.Chord((edges, hv.Dataset(nodes, "index")))

chord = chord.opts(
    opts.Chord(
        labels="date",

        # Nodes follow date order using a continuous color scale
        node_color=dim("date_order"),
        cmap="jet",
        colorbar=False,

        # Pair-linking lines are black
        edge_color="black",

        # Style
        width=650,
        height=650,
        node_size=10,
        # edge_alpha=0.75,
        # edge_line_width=2,
        title="Chord diagram of date pairs",
        hooks=[date_colorbar_hook]
    )
)

chord
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.path as mpath
from itertools import combinations

# -----------------------------
# 1. Dates
# -----------------------------
dates = [
    "2000-03-01",
    "2001-03-01",
    "2002-03-01",
    "2003-03-01",
    "2004-03-01",
    "2005-03-01",
    "2006-03-01",
    "2007-03-01",
    "2008-03-01",
    "2009-03-01",
    "2010-03-01",
]

dates = sorted(pd.to_datetime(dates))

# Labels as first day of the corresponding year
labels = [pd.Timestamp(year=d.year, month=1, day=1).strftime("%Y-%m-%d") for d in dates]

# Undirected pairs: each pair appears once
pairs = list(combinations(range(len(dates)), 2))

# -----------------------------
# 2. Circular geometry
# -----------------------------
n = len(dates)

# Start at top and go clockwise
angles = np.linspace(np.pi / 2, np.pi / 2 - 2 * np.pi, n, endpoint=False)

outer_radius = 1.00
inner_radius = 0.84
link_radius = 0.80
label_radius = 1.17

ring_width = outer_radius - inner_radius

# Colormap follows ascending date order
cmap = plt.cm.jet
colors = cmap(np.linspace(0, 1, n))

# -----------------------------
# 3. Figure
# -----------------------------
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={"aspect": "equal"})
ax.axis("off")

# -----------------------------
# 4. Draw thick circular background
# -----------------------------
step = 360 / n

for i, angle in enumerate(angles):
    center_deg = np.degrees(angle)

    wedge = patches.Wedge(
        center=(0, 0),
        r=outer_radius,
        theta1=center_deg - step / 2,
        theta2=center_deg + step / 2,
        width=ring_width,
        facecolor=colors[i],
        edgecolor="black",
        linewidth=1.5
    )
    ax.add_patch(wedge)

# -----------------------------
# 5. Draw black circular arc links
# -----------------------------
Path = mpath.Path

def polar_to_xy(radius, theta):
    return np.array([radius * np.cos(theta), radius * np.sin(theta)])

for i, j in pairs:
    theta1 = angles[i]
    theta2 = angles[j]

    p1 = polar_to_xy(link_radius, theta1)
    p2 = polar_to_xy(link_radius, theta2)

    # Mid-angle controls where the curve bends
    mid_theta = np.arctan2(
        np.sin(theta1) + np.sin(theta2),
        np.cos(theta1) + np.cos(theta2)
    )

    # Angular distance controls how deep the arc goes
    delta = np.abs(np.arctan2(np.sin(theta2 - theta1), np.cos(theta2 - theta1)))

    # Large angular separation = deeper arc toward the center
    control_radius = link_radius * (1 - 0.85 * np.sin(delta / 2))
    pc = polar_to_xy(control_radius, mid_theta)

    path = Path(
        [p1, pc, p2],
        [Path.MOVETO, Path.CURVE3, Path.CURVE3]
    )

    patch = patches.PathPatch(
        path,
        facecolor="none",
        edgecolor="black",
        linewidth=0.8,
        alpha=0.45
    )
    ax.add_patch(patch)

# -----------------------------
# 6. Draw "|" symbols instead of dots
# -----------------------------
for i, theta in enumerate(angles):
    x_inner, y_inner = polar_to_xy(inner_radius - 0.035, theta)
    x_outer, y_outer = polar_to_xy(outer_radius + 0.035, theta)

    ax.plot(
        [x_inner, x_outer],
        [y_inner, y_outer],
        color="black",
        linewidth=2.2,
        solid_capstyle="round"
    )

# -----------------------------
# 7. Add labels
# -----------------------------
for i, theta in enumerate(angles):
    x, y = polar_to_xy(label_radius, theta)

    rotation = np.degrees(theta)

    # Keep labels readable
    if rotation < -90:
        rotation += 180
        ha = "right"
    elif rotation > 90:
        rotation -= 180
        ha = "right"
    else:
        ha = "left"

    ax.text(
        x,
        y,
        labels[i],
        rotation=rotation,
        rotation_mode="anchor",
        ha=ha,
        va="center",
        fontsize=9
    )

# -----------------------------
# 8. Final layout
# -----------------------------
ax.set_xlim(-1.35, 1.35)
ax.set_ylim(-1.35, 1.35)

plt.title("Circular arc diagram of date pairs", fontsize=14, pad=20)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.path as mpath


def plot_pairs_chords(
    pairs,
    figsize=(4, 4),
    cmap="jet",
    n_color_segments=1000,

    # Circle geometry
    outer_radius=1.00,
    ring_width=0.045,
    link_radius=None,
    label_radius=1.02,

    # Year-reference "|" symbols
    show_year_ticks=True,
    year_tick_length=0.075,
    year_tick_linewidth=1.4,
    year_tick_color="black",

    # Year labels
    show_year_labels=True,
    year_label_fontsize=9,
    year_label_color="black",

    # Links
    link_color="black",
    link_linewidth=0.6,
    link_alpha=0.35,
    link_curvature=0.85,

    # Borders
    border_color="black",
    border_linewidth=1.0,

    # Layout
    xylim=1.30,
    title=None,
    ax=None,
    show=True
):
    """
    Circular arc/chord-style diagram using only already-built date pairs.

    Parameters
    ----------
    pairs : list of tuple
        Already-built date pairs.

        Example:
        [
            ("2000-03-01", "2001-04-01"),
            ("2001-04-01", "2000-03-01"),
            ("2000-03-01", "2002-05-01")
        ]

    Notes
    -----
    - The function does not create pairs internally.
    - The function infers the full date axis from the dates appearing in pairs.
    - Index pairs such as (0, 1) are not accepted because no separate date list is provided.
    """

    # -----------------------------
    # 1. Validate and convert pairs
    # -----------------------------
    if pairs is None or len(pairs) == 0:
        raise ValueError("You must provide already-built date pairs.")

    converted_pairs = []

    for pair in pairs:
        if len(pair) != 2:
            raise ValueError(f"Invalid pair: {pair}. Each pair must contain exactly two dates.")

        a, b = pair

        if isinstance(a, int) or isinstance(b, int):
            raise ValueError(
                "Index pairs are not accepted. "
                "Please provide date pairs, e.g. ('2000-03-01', '2001-04-01')."
            )

        a_date = pd.to_datetime(a)
        b_date = pd.to_datetime(b)

        converted_pairs.append((a_date, b_date))

    # -----------------------------
    # 2. Infer unique dates from pairs
    # -----------------------------
    dates = sorted(
        set(
            [a for a, b in converted_pairs] +
            [b for a, b in converted_pairs]
        )
    )

    if len(dates) < 2:
        raise ValueError("At least two unique dates are required.")

    date_to_index = {date: i for i, date in enumerate(dates)}

    indexed_pairs = [
        (date_to_index[a], date_to_index[b])
        for a, b in converted_pairs
    ]

    # -----------------------------
    # 3. Temporal circular scale
    # -----------------------------
    start_date = pd.Timestamp(year=dates[0].year, month=1, day=1)
    end_date = pd.Timestamp(year=dates[-1].year + 1, month=1, day=1)

    total_seconds = (end_date - start_date).total_seconds()

    def date_to_fraction(date):
        return (date - start_date).total_seconds() / total_seconds

    def date_to_angle(date):
        """
        January 1st of the first year starts at the top.
        Time moves clockwise.
        """
        return np.pi / 2 - 2 * np.pi * date_to_fraction(date)

    def polar_to_xy(radius, theta):
        return np.array([
            radius * np.cos(theta),
            radius * np.sin(theta)
        ])

    date_angles = [date_to_angle(d) for d in dates]

    # -----------------------------
    # 4. Geometry
    # -----------------------------
    inner_radius = outer_radius - ring_width

    if link_radius is None:
        link_radius = inner_radius + ring_width / 2

    cmap_obj = plt.get_cmap(cmap)

    # -----------------------------
    # 5. Figure
    # -----------------------------
    if ax is None:
        fig, ax = plt.subplots(figsize=figsize, subplot_kw={"aspect": "equal"})
    else:
        fig = ax.figure

    ax.axis("off")

    # -----------------------------
    # 6. Continuous circular colormap
    # -----------------------------
    for k in range(n_color_segments):
        f1 = k / n_color_segments
        f2 = (k + 1) / n_color_segments

        theta1 = np.degrees(np.pi / 2 - 2 * np.pi * f2)
        theta2 = np.degrees(np.pi / 2 - 2 * np.pi * f1)

        color = cmap_obj((f1 + f2) / 2)

        wedge = patches.Wedge(
            center=(0, 0),
            r=outer_radius,
            theta1=theta1,
            theta2=theta2,
            width=ring_width,
            facecolor=color,
            edgecolor=color,
            linewidth=0
        )

        ax.add_patch(wedge)

    # Circle borders
    ax.add_patch(
        patches.Circle(
            (0, 0),
            outer_radius,
            fill=False,
            edgecolor=border_color,
            linewidth=border_linewidth
        )
    )

    ax.add_patch(
        patches.Circle(
            (0, 0),
            inner_radius,
            fill=False,
            edgecolor=border_color,
            linewidth=border_linewidth
        )
    )

    # -----------------------------
    # 7. Draw links from provided pairs only
    # -----------------------------
    Path = mpath.Path

    for i, j in indexed_pairs:
        theta1 = date_angles[i]
        theta2 = date_angles[j]

        p1 = polar_to_xy(link_radius, theta1)
        p2 = polar_to_xy(link_radius, theta2)

        mid_theta = np.arctan2(
            np.sin(theta1) + np.sin(theta2),
            np.cos(theta1) + np.cos(theta2)
        )

        delta = np.abs(
            np.arctan2(
                np.sin(theta2 - theta1),
                np.cos(theta2 - theta1)
            )
        )

        control_radius = link_radius * (1 - link_curvature * np.sin(delta / 2))
        pc = polar_to_xy(control_radius, mid_theta)

        path = Path(
            [p1, pc, p2],
            [Path.MOVETO, Path.CURVE3, Path.CURVE3]
        )

        patch = patches.PathPatch(
            path,
            facecolor="none",
            edgecolor=link_color,
            linewidth=link_linewidth,
            alpha=link_alpha
        )

        ax.add_patch(patch)

    # -----------------------------
    # 8. Draw year-reference "|" symbols
    # -----------------------------
    years = range(dates[0].year, dates[-1].year + 1)

    tick_center_radius = inner_radius + ring_width / 2
    tick_r1 = tick_center_radius - year_tick_length / 2
    tick_r2 = tick_center_radius + year_tick_length / 2

    if show_year_ticks:
        for year in years:
            year_date = pd.Timestamp(year=year, month=1, day=1)
            theta = date_to_angle(year_date)

            x1, y1 = polar_to_xy(tick_r1, theta)
            x2, y2 = polar_to_xy(tick_r2, theta)

            ax.plot(
                [x1, x2],
                [y1, y2],
                color=year_tick_color,
                linewidth=year_tick_linewidth,
                solid_capstyle="butt"
            )

    # -----------------------------
    # 9. Add year labels only
    # -----------------------------
    if show_year_labels:
        for year in years:
            year_date = pd.Timestamp(year=year, month=1, day=1)
            theta = date_to_angle(year_date)

            x, y = polar_to_xy(label_radius, theta)

            rotation = np.degrees(theta)

            if rotation < -90:
                rotation += 180
                ha = "right"
            elif rotation > 90:
                rotation -= 180
                ha = "right"
            else:
                ha = "left"

            ax.text(
                x,
                y,
                str(year),
                rotation=rotation,
                rotation_mode="anchor",
                ha=ha,
                va="center",
                fontsize=year_label_fontsize,
                color=year_label_color
            )

    # -----------------------------
    # 10. Final layout
    # -----------------------------
    ax.set_xlim(-xylim, xylim)
    ax.set_ylim(-xylim, xylim)

    if title is not None:
        ax.set_title(title, fontsize=12, pad=15)

    if show:
        plt.show()

    return fig, ax
```

```python
def build_stepped_permutations(dates, step=1, sort_dates=True):
    """
    Build forward and backward date-pairs using a fixed maximum step.

    Parameters
    ----------
    dates : list
        Ordered list of dates or labels.

    step : int
        Maximum number of positions ahead to connect.
        For example, step=2 connects each element to the next 2 elements.

    sort_dates : bool
        If True, dates are sorted before building pairs.
        Use False if you want to preserve the original order.

    Returns
    -------
    pairs : list of tuple
        Forward and backward pairs.
    """

    if step < 1:
        raise ValueError("step must be >= 1")

    dates = list(dates)

    if sort_dates:
        dates = sorted(dates)

    pairs = []
    n = len(dates)

    for i in range(n):
        for j in range(i + 1, min(i + step + 1, n)):
            # Forward
            pairs.append((dates[i], dates[j]))

            # Backward
            pairs.append((dates[j], dates[i]))

    return pairs
```
